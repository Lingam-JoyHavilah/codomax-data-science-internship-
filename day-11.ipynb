# Day 11 - Business Insights from Employee Dataset

```python
import pandas as pd
```

```python
df = pd.read_csv("final_cleaned_employees.csv")
df
```

```python
print("Total Employees:", len(df))
print("Average Salary:", df["Salary"].mean())
print("Highest Salary:", df["Salary"].max())
print("Lowest Salary:", df["Salary"].min())
```

```python
department_count = df["Department"].value_counts()
print(department_count)
```

```python
highest_salary_employee = df.loc[df["Salary"].idxmax()]
print("Employee with Highest Salary:")
print(highest_salary_employee)
```

```python
lowest_salary_employee = df.loc[df["Salary"].idxmin()]
print("Employee with Lowest Salary:")
print(lowest_salary_employee)
```

```python
department_salary = df.groupby("Department")["Salary"].mean()
print(department_salary)
```

```python
highest_salary = df["Salary"].max()
lowest_salary = df["Salary"].min()
average_salary = df["Salary"].mean()
highest_employee = df.loc[df["Salary"].idxmax(), "Name"]
lowest_employee = df.loc[df["Salary"].idxmin(), "Name"]
largest_department = df["Department"].value_counts().idxmax()
print("Business Insights")
print("--------------------")
print(f"1. The average employee salary is {average_salary:.2f}.")
print(f"2. {highest_employee} has the highest salary of {highest_salary}.")
print(f"3. {lowest_employee} has the lowest salary of {lowest_salary}.")
print(f"4. {largest_department} has the highest number of employees.")
print(f"5. The dataset contains {len(df)} employee records.")
```

# Business Insight Report

1. The dataset provides information about employee salaries and departments.

2. The average salary gives an overall understanding of the salary level in the organization.

3. The employee with the highest salary represents the highest-paid employee in the dataset.

4. The employee with the lowest salary represents the lowest-paid employee in the dataset.

5. Department-wise employee counts help identify which department has the largest workforce.
