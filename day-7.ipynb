# Day 7 - Data Analysis using Pandas

```python
import pandas as pd
```

```python
df = pd.read_csv("cleaned_employees.csv")
df
```

```python
total_salary = df["Salary"].sum()
print("Total Salary:", total_salary)
```

```python
average_salary = df["Salary"].mean()
print("Average Salary:", average_salary)
```

```python
minimum_salary = df["Salary"].min()
print("Minimum Salary:", minimum_salary)
```

```python
maximum_salary = df["Salary"].max()
print("Maximum Salary:", maximum_salary)
```

```python
employee_count = df["Employee_ID"].count()
print("Total Employees:", employee_count)
```

```python
department_count = df["Department"].value_counts()
print(department_count)
```

```python
print("Business Insights")
print("--------------------------")
print("Total Employees :", df["Employee_ID"].count())
print("Average Salary  :", df["Salary"].mean())
print("Highest Salary  :", df["Salary"].max())
print("Lowest Salary   :", df["Salary"].min())
print("Total Salary    :", df["Salary"].sum())
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Employees",
        "Total Salary",
        "Average Salary",
        "Minimum Salary",
        "Maximum Salary"
    ],
    "Value": [
        df["Employee_ID"].count(),
        df["Salary"].sum(),
        df["Salary"].mean(),
        df["Salary"].min(),
        df["Salary"].max()
    ]
})

summary.to_csv("business_insights.csv", index=False)

print(summary)
```
