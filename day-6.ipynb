# Day 6 - Data Filtering using Pandas

```python
import pandas as pd
```

```python
df = pd.read_csv("employee.csv")
df
```

```python
print(df[["Name", "Department"]])
```

```python
it_employees = df[df["Department"] == "IT"]
print(it_employees)
```

```python
high_salary = df[df["Salary"] > 90000]
print(high_salary)
```

```python
age_filter = df[df["Age"] > 35]
print(age_filter)
```

```python
salary_sort = df.sort_values(by="Salary")
print(salary_sort)
```

```python
age_sort = df.sort_values(by="Age", ascending=False)
print(age_sort)
```

```python
result = df[(df["Department"] == "Management") & (df["Salary"] > 100000)]
print(result)
```

```python
high_salary.to_csv("filtered_employees.csv", index=False)
print("Filtered dataset saved successfully!")
```
