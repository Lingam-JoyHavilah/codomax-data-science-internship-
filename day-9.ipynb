# Day 9 - Mini Dashboard using Jupyter Notebook

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned_employees.csv")
df
```

#  Dataset Overview

```python
print(df.shape)
print(df.head())
```

# Business Insights

```python
print("Total Employees :", df["Employee_ID"].count())
print("Average Salary :", df["Salary"].mean())
print("Highest Salary :", df["Salary"].max())
print("Lowest Salary :", df["Salary"].min())
print("Total Salary :", df["Salary"].sum())
```

# Department-wise Employee Count

```python
department_count = df["Department"].value_counts()
print(department_count)
```

# Employee Salary Bar Chart

```python
plt.figure(figsize=(10,5))
plt.bar(df["Name"], df["Salary"])
plt.title("Employee Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Salary Trend

```python
plt.figure(figsize=(10,5))
plt.plot(df["Name"], df["Salary"], marker="o")
plt.title("Salary Trend")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Department Distribution

```python
plt.figure(figsize=(7,7))
plt.pie(
    department_count,
    labels=department_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Department Distribution")
plt.show()
```
