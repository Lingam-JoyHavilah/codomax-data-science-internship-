#  Employee Data Analysis Dashboard

### Data Science Internship – Day 12

This notebook presents an analysis of employee data using Python, Pandas, and Matplotlib.

**Objectives:**
- Explore the employee dataset
- Generate business insights
- Analyze employee salaries and departments
- Visualize data using charts

```python
import pandas as pd
import matplotlib.pyplot as plt
```

##  1. Load and Explore the Dataset

First, we load the cleaned employee dataset and inspect the available records.

```python
df = pd.read_csv("final_cleaned_employees.csv")
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

##  2. Business Insights

The following calculations provide a quick summary of the employee dataset.

```python
total_employees = df["Employee_ID"].count()
average_salary = df["Salary"].mean()
highest_salary = df["Salary"].max()
lowest_salary = df["Salary"].min()
total_salary = df["Salary"].sum()
print("Total Employees :", total_employees)
print("Average Salary  :", round(average_salary, 2))
print("Highest Salary  :", highest_salary)
print("Lowest Salary   :", lowest_salary)
print("Total Salary    :", total_salary)
```

##  3. Department-wise Employee Distribution

This section shows the number of employees in each department.

```python
department_count = df["Department"].value_counts()
department_count
```

## 4. Employee Salary Comparison

The bar chart compares the salaries of individual employees.

```python
plt.figure(figsize=(10, 5))
plt.bar(df["Name"], df["Salary"])
plt.title("Employee Salary Comparison", fontsize=14)
plt.xlabel("Employee Name")
plt.ylabel("Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 5. Salary Trend

The line chart shows the variation in employee salaries across the dataset.

```python
plt.figure(figsize=(10, 5))
plt.plot(df["Name"], df["Salary"], marker="o")
plt.title("Employee Salary Trend", fontsize=14)
plt.xlabel("Employee Name")
plt.ylabel("Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##  6. Department Distribution

The pie chart represents the proportion of employees across different departments.

```python
plt.figure(figsize=(7, 7))
plt.pie(
    department_count,
    labels=department_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Employee Distribution by Department", fontsize=14)
plt.show()
```

##  7. High Salary Employees

The following table displays employees whose salary is greater than 90,000.

```python
high_salary = df[df["Salary"] > 90000]
high_salary
```

##  8. Key Observations

Based on the analysis, the following observations can be made:

- The dataset contains multiple employee records across different departments.
- Salary statistics provide an overview of the organization's employee compensation.
- Employees are distributed across different departments.
- The salary charts make it easier to compare employee compensation.
- Filtering helps identify employees whose salary is above the selected threshold.
